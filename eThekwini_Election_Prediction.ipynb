### Important Treatment Decision

Election counts such as **RegisteredVoters, SpoiltVotes and TotalValidVotes should not be negative**. Any negative values are therefore flagged as invalid and converted to missing values rather than silently being accepted.

The original dataset is preserved, while a separate cleaned dataset is created for analysis and modelling. This ensures that all data-cleaning decisions remain transparent and reproducible.


````python

# DATA PREPARATION & VALIDATION
# eThekwini 2021 Local Government Election final check up 


# Keep the original dataset unchanged
clean_df = df.copy()



# 1. Dataset structure


print("Original dataset shape:")
print(clean_df.shape)

print("\nData types:")
display(clean_df.dtypes)



# 2. Check duplicate records


duplicates = clean_df.duplicated().sum()

print("\nNumber of duplicate rows:")
print(duplicates)

# Remove exact duplicate rows

clean_df = clean_df.drop_duplicates()



# 3. Check missing values


print("\nMissing values:")
display(clean_df.isnull().sum())



# 4. Convert election variables to numeric


numeric_columns = [
    "RegisteredVoters",
    "SpoiltVotes",
    "TotalValidVotes"
]

for col in numeric_columns:
    clean_df[col] = pd.to_numeric(
        clean_df[col],
        errors="coerce"
    )



# 5. Check for negative election values


negative_counts = {}

for col in numeric_columns:
    negative_counts[col] = (
        clean_df[col] < 0
    ).sum()

print("\nNegative values detected:")
display(pd.Series(negative_counts))



# 6. Convert suspicious negative values to NaN


for col in numeric_columns:
    clean_df.loc[
        clean_df[col] < 0,
        col
    ] = np.nan



# 7. Check important categorical variables


print("\nNumber of unique wards:")
print(clean_df["Ward"].nunique())

print("\nNumber of voting districts:")
print(clean_df["VotingDistrict"].nunique())

print("\nNumber of political parties:")
print(clean_df["PartyName"].nunique())

print("\nBallot types:")
display(clean_df["BallotType"].value_counts())


# 8. Standardise relevant text fields


text_columns = [
    "Municipality",
    "Ward",
    "VotingDistrict",
    "VotingStationName",
    "BallotType",
    "PartyName"
]

for col in text_columns:
    clean_df[col] = (
        clean_df[col]
        .astype("string")
        .str.strip()
    )



# 9. Validate essential modelling fields


essential_columns = [
    "Ward",
    "PartyName",
    "TotalValidVotes"
]

missing_essential = clean_df[
    clean_df[essential_columns].isnull().any(axis=1)
]

print("\nRows missing essential modelling information:")
print(len(missing_essential))



# 10. Final missing-value summary


print("\nFinal missing-value summary:")
display(clean_df.isnull().sum())



# 11. Final dataset shape


print("\nCleaned dataset shape:")
print(clean_df.shape)
````

## 9.Time Series Forecasting

We use this to predict what it may happen base on the data we have for the past

````python


import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Make a copy
ts_df = clean_df.copy()

# Convert date column
ts_df["DateGenerated"] = pd.to_datetime(
    ts_df["DateGenerated"],
    errors="coerce"
)

# Remove invalid dates and missing votes
ts_df = ts_df.dropna(
    subset=["DateGenerated", "TotalValidVotes"]
)

# Group votes by date
daily_votes = (
    ts_df.groupby("DateGenerated")["TotalValidVotes"]
    .sum()
    .sort_index()
)

print("Historical election data:")
display(daily_votes)
````
